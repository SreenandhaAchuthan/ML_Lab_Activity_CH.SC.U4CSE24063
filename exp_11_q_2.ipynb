```python
#import pandas
import pandas as pd
```

```python
#table
data = {
    'Variable': [
        'Climate',
        'Housing',
        'Health',
        'Crime',
        'Transportation',
        'Education',
        'Arts',
        'Recreation',
        'Economy'
    ],

    'PC1': [
        0.190,
        0.544,
        0.782,
        0.365,
        0.585,
        0.394,
        0.985,
        0.520,
        0.142
    ],

    'PC2': [
        0.017,
        0.020,
        -0.605,
        0.294,
        0.085,
        -0.273,
        0.126,
        0.402,
        0.150
    ],

    'PC3': [
        0.207,
        0.204,
        0.144,
        0.585,
        0.234,
        0.027,
        -0.111,
        0.519,
        0.239
    ]
}

df = pd.DataFrame(data)

df
```

```python
#highest loading for PC1
pc1_variable = df.loc[df['PC1'].abs().idxmax(), 'Variable']
pc1_value = df['PC1'].abs().max()

print("PC1:", pc1_variable)
print("Loading:", pc1_value)
```

```python
#highest loading for pc2
pc2_variable = df.loc[df['PC2'].abs().idxmax(), 'Variable']
pc2_value = df['PC2'].abs().max()

print("PC2:", pc2_variable)
print("Loading:", pc2_value)
```

```python
#highest possible for pc3
pc3_variable = df.loc[df['PC3'].abs().idxmax(), 'Variable']
pc3_value = df['PC3'].abs().max()

print("PC3:", pc3_variable)
print("Loading:", pc3_value)
```

```python
#all 3
for pc in ['PC1', 'PC2', 'PC3']:

    index = df[pc].abs().idxmax()

    variable = df.loc[index, 'Variable']
    loading = df.loc[index, pc]

    print(
        pc,
        "->",
        variable,
        "with loading",
        loading
    )
```

### Notebook Summary

**What we did:**
We analyzed principal component loadings across nine different quality-of-life variables (like Climate, Housing, and Health) to identify which variables have the strongest relationship with the first three principal components (PC1, PC2, and PC3).

**How we did it:**
1. **Data Setup:** Loaded the loading values into a Pandas DataFrame.
2. **Analysis:** Used Pandas to find the absolute maximum loading values for each principal component (`.abs().idxmax()`).
3. **Key Findings:**
   * **PC1** is dominated by **Arts** (loading: 0.985).
   * **PC2** is dominated by **Health** (loading: -0.605).
   * **PC3** is dominated by **Crime** (loading: 0.585).

```python
import math
from collections import Counter

# Existing data
data = [
    (40, 20, "Red"),
    (50, 50, "Blue"),
    (60, 90, "Blue"),
    (10, 25, "Red"),
    (70, 70, "Blue"),
    (60, 10, "Red"),
    (25, 80, "Blue")
]

# New entry
new_entry = (50, 40)

# K value
k = 3

# Calculate distances
distances = []

for brightness, saturation, class_name in data:

    distance = math.sqrt(
        (new_entry[0] - brightness)**2 +
        (new_entry[1] - saturation)**2
    )

    distances.append((distance, class_name))

# Sort distances
distances.sort()

# Select K nearest
nearest = distances[:k]

print("K nearest neighbours:")

for distance, class_name in nearest:
    print(class_name, "Distance =", round(distance, 2))

# Majority voting
classes = [item[1] for item in nearest]

prediction = Counter(classes).most_common(1)[0][0]

print("\nPredicted class:", prediction)
```

### Summary of this Notebook

**What we did:**
We predicted the color class ("Red" or "Blue") for a new data point `(50, 40)` using the **K-Nearest Neighbors (KNN)** algorithm.

**How we did it:**
1. **Calculated Distance:** Used the Euclidean distance formula ($d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}$) to find the distance between our new entry and every existing data point based on brightness and saturation.
2. **Sorted & Selected:** Sorted all calculated distances in ascending order and picked the $K=3$ closest data points (neighbors).
3. **Majority Vote:** Counted the classes of these 3 nearest neighbors. Because "Red" appeared twice and "Blue" appeared once, "Red" won the majority vote, making it our final predicted class.

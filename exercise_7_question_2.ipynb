```python
#libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
```

```python
#dataset creation
data = {
    'Weather': [
        'Sunny', 'Rainy', 'Sunny', 'Rainy',
        'Rainy', 'Sunny', 'Rainy', 'Sunny',
        'Sunny', 'Rainy', 'Sunny', 'Rainy'
    ],

    'Decision': [
        'Go out', 'Stay in', 'Go out', 'Stay in',
        'Stay in', 'Go out', 'Stay in', 'Go out',
        'Go out', 'Stay in', 'Go out', 'Stay in'
    ]
}

df = pd.DataFrame(data)
print("original dataset:")
print(df)
```

```python
#convert category values to numbers
df['Weather'] = df['Weather'].map({'Sunny':0, 'Rainy':1})
df['Decision'] = df['Decision'].map({'Stay in':0,'Go out':1})
```

```python
#sepaerate feature and target
X = df[['Weather']]
y = df[['Decision']]
```

```python
#seperate training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X,y,
    test_size=0.25,
    random_state= 42,
    stratify=y
)
```

```python
#desicion tree classifier
model = DecisionTreeClassifier(max_depth=1,random_state=42)
```

```python
#train
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
#accuracy
accuracy = accuracy_score(y_test,y_pred)
print("\nActual values:   ", y_test.values)
print("Predicted values:", y_pred)
print("\nAccuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")
```

```python
#confusion matrix
cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Stay in', 'Go out']
)

disp.plot()
plt.show()
```

```python
plt.figure(figsize=(8, 8))

plot_tree(
    model,
    feature_names=['Weather'],
    class_names=['Stay in', 'Go out'],
    filled=True
)

plt.show()
```

### Summary of this Notebook

#### What this notebook shows:
It shows how to build and evaluate a basic Decision Tree Classifier that predicts whether a person will go out or stay in based on the weather.

#### What we did:
1. **Created a small dataset** mapping Weather ('Sunny', 'Rainy') to a Decision ('Go out', 'Stay in').
2. **Prepared the data** by converting categorical text into numbers (0 and 1) and splitting it into training (75%) and testing (25%) sets.
3. **Trained a Decision Tree** with a maximum depth of 1.
4. **Evaluated the model** using accuracy, a confusion matrix, and a visual plot of the decision tree itself.

#### How we did it:
* Used **pandas** to create and map the dataset.
* Used **scikit-learn** to split the data (`train_test_split`), train the model (`DecisionTreeClassifier`), and calculate metrics (`accuracy_score`, `confusion_matrix`).
* Used **matplotlib** to plot the confusion matrix and visualize the split rules of the decision tree.
